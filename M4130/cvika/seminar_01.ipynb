# Bonus

<div class="alert alert-block alert-success">

Student si během procvičování ukázkových testů zaznamenával počet bodů, které získal. K úspěšnému absolvování testu potřebuje získat alespoň 50 bodů. Jaká byla jeho nejdelší série úspěšných testů v řadě?

Zkuste problém vyřešit jen za pomocí znalostí z tohoto cvičení a přednášky. Nepoužívejte `for` cykly!

````r
test_results <- c(12, 24, 51, 33, 67, 21, 55, 20, 70, 55, 49, 81, 71, 51, 50, 48, 91, 88, 77, 50, 66, 64, 64, 82, 97)

test_results_zero <- c(0, as.integer(test_results > 49), 0)

test_results_diff <- diff(test_results_zero)

positions <- 1:length(test_results_diff)

combo <- test_results_diff * positions

clean_combo <- combo[combo != 0]

plus <- rep(c(1, -1), length.out = length(clean_combo))

plus_combo <- clean_combo * plus

results <- diff(plus_combo)

max(results)

````
