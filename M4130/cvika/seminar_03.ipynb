Data **fitness**, se kterými budeme dnes pracovat, obsahují výsledky fyzických testů 12 armádních kandidátů. Dimenze datového pole jsou následující:

1. číslo kandidáta (1-6)
2. disciplína (kliky, sedy-lehy, běh)
3. pohlaví

Pole **limits** obsahuje bodové hranice pro hodnocení fyzických testů. Struktura je:

1. počet bodů (1-5)
2. disciplína (kliky, sedy-lehy, běh)
3. pohlaví

````r
fitness <- array(
    c(
        # MEN - push-ups
        58, 44, 35, 50, 28, 62,
        # MEN - sit-ups
        52, 46, 39, 41, 35, 55,
        # MEN - 2 km run in minutes
        8.6, 9.4, 10.2, 9.0, 11.1, 8.3,
        # WOMEN - push-ups
        42, 31, 37, 25, 46, 34,
        # WOMEN - sit-ups
        50, 45, 48, 36, 54, 44,
        # WOMEN - 2 km run in minutes
        9.2, 10.0, 9.6, 10.8, 8.9, 9.7
    ),
    dim = c(6, 3, 2),
    dimnames = list(
        candidate = paste0("candidate_", 1:6),
        discipline = c("push_ups", "sit_ups", "run_2km"),
        group = c("men", "women")
    )
)
print(fitness)
````

````r
limits <- array(c(
        # MEN - push-ups
        25, 32, 40, 48, 55,
        # MEN - sit-ups
        30, 36, 42, 48, 54,
        # MEN - run
        12, 11, 10, 9, 8.5,
        # WOMEN - push-ups
        15, 22, 28, 35, 42,
        # WOMEN - sit-ups
        28, 34, 40, 46, 52,
        # WOMEN - run
        12.5, 11.5, 10.5, 9.5, 9
    ),
    dim = c(5, 3, 2),
    dimnames = list(
        points = 1:5,
        discipline = c("push_ups", "sit_ups", "run_2km"),
        group = c("men", "women")
    )
)
print(limits)
````
