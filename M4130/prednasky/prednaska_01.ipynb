# .ipynb

<div class="alert alert-block alert-info"> 

**Jupyter Notebook** (.ipynb) je typ souboru, který umožňuje psát text a kód v jednom dokumentu. 

Kód lze spouštět po jednotlivých částech (buňkách) a výsledky se zobrazují přímo pod příslušnou buňkou. 

Kromě zdrojového kódu lze do notebooku vkládat také obrázky, odkazy, tabulky, matematické vzorce nebo přehledně formátovaný text. Díky tomu je snadné experimentovat s kódem, průběžně si ověřovat výsledky a doplňovat vysvětlující poznámky. 

</div>

## Typy buňek

<div class="alert alert-block alert-info"> 

Jupyter Notebook obsahuje dva základní typy buněk:

1. **code** – slouží pro psaní a spouštění kódu
2. **markdown** – slouží pro psaní textu, nadpisů, seznamů, obrázků, odkazů a další dokumentace

</div>

## Orientace

<div class="alert alert-block alert-info"> 

Pro lepší orientaci je vhodné notebook rozdělit do **kapitol a podkapitol** pomocí nadpisů. Nadpisy se zapisují v **markdown** buňkách pomocí znaku `#`. Jeden znak `#` označuje hlavní kapitolu, dva znaky `##` podkapitolu, tři znaky `###` podkapitolu nižší úrovně atd.

Například:

```markdown
# Analýza dat

## Načtení dat

## Čištění dat

### Odstranění chybějících hodnot

### Odstranění odlehlých hodnot

## Vizualizace výsledků
```

Jupyter z těchto nadpisů automaticky vytváří **Outline** (obsah dokumentu), který zobrazuje strukturu notebooku. Kliknutím na libovolný nadpis v Outline lze rychle přejít na příslušnou část dokumentu.

```text
Analýza dat
├── Načtení dat
├── Čištění dat
│   ├── Odstranění chybějících hodnot
│   └── Odstranění odlehlých hodnot
└── Vizualizace výsledků
```

</div>

## Klávesové zkratky

<div class="alert alert-block alert-info"> 

Při práci v Jupyter Notebooku ve Visual Studio Code lze řadu činností provádět pomocí klávesových zkratek. Notebook pracuje ve dvou režimech:

- **edit mode** – slouží k úpravě obsahu dané buňky
- **command mode** – slouží ke správě buněk

Mezi režimy lze přepínat:

- **Enter** – přepne do režimu úprav aktuální buňky (edit mode)
- **Esc** – ukončí úpravu buňky a přepne do režimu příkazů (command mode)

V **command mode** jsou užitečné například tyto zkratky:

| Klávesa | Akce |
|---------|------|
| ↑ / ↓ | Přesun na předchozí nebo následující buňku |
| **A** | Vloží novou buňku nad aktuální |
| **B** | Vloží novou buňku pod aktuální |
| **X** | Vyjme aktuální buňku |
| **C** | Zkopíruje aktuální buňku |
| **V** | Vloží zkopírovanou nebo vyjmutou buňku pod aktuální |

</div>

# Data

<div class="alert alert-block alert-info">

Při programování v R je důležité rozlišovat **datové typy**, **datové struktury** a **funkce**. Každý z těchto pojmů označuje něco jiného.

- **Datový typ** určuje, **jaký druh hodnoty je uložen v jedné proměnné**.
- **Datová struktura** určuje, **jak jsou jednotlivé hodnoty uspořádány**.
- **Funkce** představuje **blok kódu**, který provede určitou činnost. Funkce může přijímat vstupní hodnoty (argumenty) a vracet výsledek.
Argumenty jsou vstupní hodnoty, se kterými funkce pracuje.

</div>

<div class="alert alert-block alert-info">

## Datové typy 

| Datový typ (R název) | Co reprezentuje | Příklad v R |
| :--- | :--- | :--- |
| **`numeric`** (nebo `double`) | Reálná čísla | `x_numeric <- 3.14` nebo `y_double <- 42` |
| **`integer`** | Celá čísla (zapisují se s příponou `L`) | `x_integer <- 5L` |
| **`character`** | Textové řetězce (zápis v uvozovkách) | `x_character <- "Hello world!"` |
| **`logical`** | Logické hodnoty (pravda / nepravda) | `x_logical <- TRUE` nebo `y_logical <- F` |
| **`complex`** | Komplexní čísla | `x_complex <- 1 + 4i` |
| **`raw`** | Surová data v bajtech (používá se zřídka) | `x_raw <- charToRaw("A")` |

<div class="alert alert-block alert-info">

## Logické proměnné

<div class="alert alert-block alert-info">

Používejte radějí `TRUE` a `FALSE` než pouze jejich zkratky `T` a `F`.

</div>

<div class="alert alert-block alert-success">

Vyzkoušejte zadefinovat proměnné z předchozí tabulky.

</div>

````r
x_numeric <- 3.14
y_double <- 42
x_integer <- 5L
x_character <- "Hello World!"
x_logical <- TRUE
y_logical <- FALSE
x_complex <- 1 + 4i
x_raw <- charToRaw("Hello World!")
````

<div class="alert alert-block alert-danger">

Do proměnné **T** je zakódována binární hodnota 1, čili **TRUE**. Naopak do proměnné **F** je zakódováno **FALSE**, neboli logická nula. Proto nikdy nenazývejte proměnné **F** a **T**.

<div class="alert alert-block alert-info">

Datový typ proměnné můžeme vypsat pomocí příkazů **typeof**, **str**. Případně pro některé typy ještě dávají smysl příkazy **mode**, **class** nebo **storage.mode**.

</div>

<div class="alert alert-block alert-info">

| Funkce | Úroveň pohledu | Co zjišťuje | Příklad pro `1.5` vs `1L` |
| :--- | :--- | :--- | :--- |
| **`typeof()`** | Nízkoúrovňová (C) | Přesný typ uložení v paměti | `double` vs `integer` |
| **`class()`** | Vysokoúrovňová (S3/S4) | Objektovou třídu a chování funkcí | `numeric` vs `integer` |
| **`str()`** | Strukturální diagnostika | Kompletní přehled, rozměry a ukázku dat | `num 1.5` vs `int 1` |
| **`mode()`** | Konceptuální (jazyk S) | Základní teoretickou kategorii | `numeric` vs `numeric` |
| **`storage.mode()`** | Rozhraní (C/Fortran) | Paměťový formát pro externí kód | `double` vs `integer` |

</div>

<div class="alert alert-block alert-success">

Vyzkoušejte použití těchto příkazů na proměnných z předchozího příkladu.

</div>

````r
z <- x_numeric
typeof(z)
str(z)
mode(z)
class(z)
storage.mode(z)
````

Funkce se volají dvěma způsoby:

Symbolu `|>` se říká **pipe** a používá se pro řetězení příkazů. Můžete ho číst jako "a potom". Je vhodný zejména v případech, kdy voláme více funkcí po sobě.

````r
x_numeric |> typeof() |> cat()
````
