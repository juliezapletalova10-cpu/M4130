# 5. přednáška

## Knihovny v R

<div class="alert alert-block alert-info"> 

**CRAN** (*Comprehensive R Archive Network*) je oficiální centrální repozitář pro jazyk R. Slouží jako globální síť webových serverů, které ukládají a distribuují samotný softwarový balík R, dokumentaci a především tisíce rozšíření (balíčků/knihoven).
</div>

<div class="alert alert-block alert-info"> 

#### Vlastnosti CRANu:

* **Přísná kontrola kvality:** Každý balíček projde před publikací automatickými i manuálními testy. Musí obsahovat funkční kód, dokumentaci a příklady bez chyb.
* **Automatické řešení závislostí:** Při instalaci balíčku přes `install.packages()` CRAN automaticky doinstaluje i všechny ostatní knihovny, které požadovaný balíček potřebuje k fungování.
* **Verzování a archivace:** CRAN udržuje vždy nejnovější stabilní verze balíčků, ale zároveň uchovává i jejich starší verze pro zpětnou kompatibilitu.
* **Garantovaná dostupnost:** Jedná se o primární a nejbezpečnější zdroj knihoven pro R (na rozdíl od neověřených kódů např. z GitHubu).
</div>

### 1. Instalace knihovny


<div class="alert alert-block alert-info"> 

Knihovnu stačí nainstalovat z oficiálního repozitáře CRAN **pouze jednou**. Instaluje se pomocí příkazu `install.packages("nazev_knihovny")`
</div>

### 2. Načtení knihovny


<div class="alert alert-block alert-info"> 

Před každým použitím v novém skriptu nebo relaci musíte knihovnu **načíst do paměti** příkazem `library(nazev_knihovny)`
</div>

<div class="alert alert-block alert-warning"> 

Všimněte si zadání uvozovek v předchozích dvou příkazech.
</div>

### 3. Použití funkce bez načtení celé knihovny


<div class="alert alert-block alert-info"> 

Pokud potřebujete použít jen jednu konkrétní funkci, lze ji zavolat přímo pomocí operátoru `::` bez předchozího načtení balíčku přes `library()`, tj. ve formátu `nazev_knihovny::prikaz(parametry)`. Tento přístup zároveň řeší případné konflikty, kdy dvě různé knihovny obsahují funkce se stejným názvem.

</div>

### 4. Užitečné příkazy pro správu


<div class="alert alert-block alert-info"> 

- Aktualizace balíčků: `update.packages()`

- Nápověda ke knihovně: `help(package = "nazev_knihovny")`

- Seznam nainstalovaných balíčků: `installed.packages()`
</div>

## Dokumentace ke knihovnám


<div class="alert alert-block alert-info"> 

#### 1. README soubor
Podobně jako u jiných programovacích jazyků slouží `README.md` (nebo `README.Rmd`) jako hlavní vizitka balíčku v repozitáři.

* **Obsah:** Stručný popis, co balíček dělá, návod k instalaci a krátký ukázkový příklad použití.
* **Rozšíření v R (`README.Rmd`):** V R se často používá knitr/rmarkdown s příponou `.Rmd`. Tento soubor umožňuje vkládat živý R kód, který se při vyrenderování do `README.md` automaticky spustí a vloží aktuální výstupy či grafy.


#### 2. DESCRIPTION soubor
Pro definici závislostí má každý R balíček v kořenovém adresáři soubor `DESCRIPTION`. Závislosti se zapisují do specifických sekcí:

* **`Imports:`** Balíčky, které váš kód nutně potřebuje k fungování (nainstalují se automaticky).
* **`Suggests:`** Balíčky potřebné jen pro testování, tvorbu dokumentace nebo volitelné funkce.
* **`Depends:`** Starší způsob definování závislostí (dnes se preferuje `Imports`).

*Ukázka v souboru `DESCRIPTION`:*

```text
Imports:
    dplyr (>= 1.0.0),
    ggplot2,
    stringr
Suggests:
    testthat
```
</div>

<div class="alert alert-block alert-success"> 

Jak vytvořit nový sloupec *p.ratio*, který bude popisovat poměr výkon/objem a sloupec *ecol.ratio*, který bude popisovat poměr výkon/spotřeba?
    
</div>

<div class="alert alert-block alert-danger"> 

Příkaz **mutate** mění některé proměnné nebo vytváří nové a přidává je do tabulky.
    
</div>

<div>
<img src="https://raw.githubusercontent.com/JanKolacek/figs/refs/heads/main/mutate.png" width="500"/>

````r
mtcars |> mutate(p.ratio = hp / disp, ecol.ratio = hp / mpg)
````

<div class="alert alert-block alert-success"> 

Jak vytvořit totéž, ale jen pro auta s výkonem větším než 200 koní?
    
</div>

````r
mtcars |>  filter(hp > 200)  |>  mutate(p.ratio = hp / disp, ecol.ratio = hp / mpg)
````

<div class="alert alert-block alert-success"> 

Přejmenujte sloupec *hp* na *horse_power* pomocí **rename**.
    
</div>

````r
mtcars  |> rename(horse_power = hp)
````

<div class="alert alert-block alert-success"> 

Vytvořte nový sloupec *kategorie_vykonu*, který bude mít hodnoty *Weak* pro výkon menší než 100, *Medium* pro výkon mezi 100 a 200 a *Strong* pro výkon větší než 200. Použijte funkci **case_when**.
    
</div>

````r
mtcars  |>  mutate(kategorie_vykonu = case_when(hp < 100 ~ "weak", between(hp, 100, 200) ~ "medium", TRUE ~ "strong"))
````

<div class="alert alert-block alert-success"> 

Jak zjistit průměrné *ecol.ratio* pro automobily s příslušným počtem válců?
    
</div>

<div class="alert alert-block alert-danger"> 

Parametr **.by** rozdělí data do příslušných skupin a provede se operace v těchto skupinách.
    
</div>

````r
mtcars |> mutate(ecol.ratio = hp / mpg) |> mutate(mean.ecol.ratio = mean(ecol.ratio), .by = cyl)  |> select(cyl, ecol.ratio, mean.ecol.ratio)
````

<div class="alert alert-block alert-danger"> 

Alternativa k **.by** je příkaz **group_by** v kombinaci s **ungroup**.
    
</div>

````r
mtcars |> mutate(ecol.ratio = hp / mpg) |> 
    group_by(cyl) |> 
    mutate(mean.ecol.ratio = mean(ecol.ratio))  |> 
    ungroup() |> 
    select(cyl, ecol.ratio, mean.ecol.ratio)
````
