## Typická datová analýza v R


<div class="alert alert-block alert-info"> 

### 1. Jak vypadá typický životní cyklus datové analýzy

Datová analýza v R obvykle sleduje ucelený workflow (často vycházející z metodiky Tidyverse):

1. **Načtení a import dat:** Načtení dat z CSV, Excelu, databází nebo API (`readr`, `readxl`, `fromJSON`).
2. **Čištění a úprava:** Ošetření chybějících hodnot (`NA`), filtrace, tvorba nových sloupců a agregace (`dplyr`, `tidyr`).
3. **Průzkumová analýza a vizualizace:** Souhrnné statistiky, objevování vzorců a tvorba grafů (`ggplot2`).
4. **Modelování a statistika:** Regrese, klasifikace, testování hypotéz (`stats`, `tidymodels`).
5. **Export a prezentace:** Uložení výsledků do databáze/souboru, generování reportů (**Quarto**, **R Markdown**) nebo výstupů pro **Jupyter**.
</div>

<div class="alert alert-block alert-info"> 

### 2. Tvorba kapitol a sekcí ve skriptu (Document Outline)

Ve VS Code můžete R skript rozdělit na přehledné logické kapitoly. Rozhraní automaticky rozpozná speciální komentáře a vygeneruje z nich stromovou strukturu kódu (Outline).

Sekci v R vytvoříte tak, že za komentářový řádek přihodíte **nejméně 4 pomlčky (`----`), křížky (`####`) nebo rovnitka (`====`)**:

</div>

```R
# 1. NAČTENÍ KNIHOVEN A DAT ----
library(tidyverse)
library(data.table)

df <- fread("data.csv")


# 2. ČIŠTĚNÍ DAT ====
df_clean <- df %>%
  filter(!is.na(customer_id)) %>%
  mutate(amount = ifelse(amount < 0, 0, amount))


# 3. VIZUALIZACE A EDA ####
ggplot(df_clean, aes(x = category, y = amount)) +
  geom_boxplot()
```

### 4. Rozdělení kódu na více částí

Místo jednoho „obřího“ skriptu o tisících řádcích je osvědčenou praxí kód rozdělit do více menších souborů podle jejich účelu:

#### A. Struktura složek projektu
</div>

```text
muj_projekt/
├── 01_import.R          # Načtení dat z externích zdrojů
├── 02_cisteni.R         # Transformace a úprava typů
├── 03_vizualizace.R     # Grafy a reportování
├── R/
│   └── pomocne_funkce.R # Vlastní R funkce (Custom Functions)
└── main.R               # Hlavní skript, který spouští celý proces
```
<div class="alert alert-block alert-info"> 

#### B. Propojení skriptů pomocí **source**
V hlavním skriptu (`main.R`) nebo Jupyter notebooku pak jednotlivé části postupně spouštíte:

</div>

```R
# Načtení vlastních funkcí
source("R/pomocne_funkce.R")

# Postupné spuštění pipeline
source("01_import.R")
source("02_cisteni.R")
source("03_vizualizace.R")
```

**19.** Jak uložit data do proměnné?

````r
nazev_filmu <- "Pelíšky"
jmena_hercu <- c("Miroslav Donutil", "Jiří Kodet", "Simona Stašová", 
                 "Emília Vášáryová", "Bolek Polévka", "Jaroslav Dušek", "Eva Holubová", "Stella Zázvorková")
skore_recenze <- c(4, 3.5, 5)
zdroj_recenze <- c("csfd1", "csfd2", "csfd3")
komentare_recenze <- c("Tenhle film mám rád, byť rozhodně není Hřebejkovým nejlepším.",
                       "Nepochopitelným způsobem nadhodnocený film!", 
                       "Málokdy se vidí, aby se z něčeho stala skoro ihned klasika.")
rok_filmu <- 1999

recenze_filmu <- data.frame(skore = skore_recenze, zdroj = zdroj_recenze, komentar = komentare_recenze)
# vytvorime seznam
film <- list(nazev = nazev_filmu, herci = jmena_hercu, recenze = recenze_filmu, rok = rok_filmu)

film
````
