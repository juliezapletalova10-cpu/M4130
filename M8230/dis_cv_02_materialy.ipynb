# M8230: Posloupnosti, posun, diference a monotonie

Na tabuli jsme nejprve ručně počítali vybrané příklady a zopakovali si základní práci s diferencemi.

Tento notebook na ruční část **obsahově navazuje**, ale neslouží pouze ke kontrole stejných výpočtů. Pomocí Pythonu se zaměříme hlavně na to, co lze z hodnot, diferencí a grafů posloupnosti **vidět a interpretovat**.

Budeme používat zejména:

- `np.diff(a)` pro první diferenci posloupnosti,
- `np.diff(a, 2)` pro druhou diferenci,
- `np.diff(a, k)` pro diferenci řádu $k$,
- grafy pro porovnání posloupnosti a jejích diferencí.

Připomeňme pouze

$$
Ea(t)=a(t+1),
$$

$$
\Delta a(t)=a(t+1)-a(t).
$$

Znaménko první diference je důležité pro monotonii:

$$
\Delta a(t)>0 \Rightarrow a(t+1)>a(t),
$$

$$
\Delta a(t)<0 \Rightarrow a(t+1) < a(t).
$$


```r
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from math import comb
```

## 1. Čtyři posloupnosti: co nám řekne první diference?

V počítačové části budeme nejprve pracovat se čtyřmi posloupnostmi:

$$
\begin{aligned}
\text{a)}\quad &a(t)=3t-5,\\
\text{b)}\quad &a(t)=t^2-4t+1,\\
\text{c)}\quad &a(t)=2^{-t},\\
\text{d)}\quad &a(t)=(-1)^t(t+1).
\end{aligned}
$$

U každé vypíšeme několik prvních hodnot a první diferenci. Potom porovnáme, co lze vyčíst z grafu posloupnosti a co ze znaménka její diference.


```r
t = np.arange(0, 9)

a1 = 3*t - 5
a2 = t**2 - 4*t + 1
a3 = 2.0**(-t)
a4 = (-1.0)**t * (t + 1)

D1 = np.diff(a1)
D2 = np.diff(a2)
D3 = np.diff(a3)
D4 = np.diff(a4)

print(f"t = {t}")

print()
print(f"a) a(t) = {a1}")
print(f"   Δa(t) = {D1}")

print()
print(f"b) a(t) = {a2}")
print(f"   Δa(t) = {D2}")

print()
print(f"c) a(t) = {a3}")
print(f"   Δa(t) = {D3}")

print()
print(f"d) a(t) = {a4.astype(int)}")
print(f"   Δa(t) = {D4.astype(int)}")
```

### Grafy posloupností

Nejprve se podíváme na samotné hodnoty posloupností. Zkuste si ještě před vykreslením tipnout, která z nich bude růst, klesat, měnit směr nebo oscilovat.


```r
plt.figure(figsize=(9, 5))
plt.plot(t, a1, marker="o", label="a) 3t - 5")
plt.plot(t, a2, marker="o", label="b) t² - 4t + 1")
plt.plot(t, a3, marker="o", label="c) 2⁻ᵗ")
plt.plot(t, a4, marker="o", label="d) (-1)ᵗ(t+1)")
plt.axhline(0, linestyle="--")
plt.xlabel("t")
plt.ylabel("a(t)")
plt.title("Čtyři posloupnosti")
plt.grid(True)
plt.legend()
plt.show()
```

Nyní vykreslíme první diference. Hodnota $\Delta a(t)$ popisuje změnu mezi časy $t$ a $t+1$.

Při čtení grafu sledujte především **znaménko diference** a místa, kde se znaménko mění.


```r
t_diff = t[:-1]

plt.figure(figsize=(9, 5))
plt.plot(t_diff, D1, marker="o", label="a) Δa(t)")
plt.plot(t_diff, D2, marker="o", label="b) Δa(t)")
plt.plot(t_diff, D3, marker="o", label="c) Δa(t)")
plt.plot(t_diff, D4, marker="o", label="d) Δa(t)")
plt.axhline(0, linestyle="--")
plt.xlabel("t")
plt.ylabel("Δa(t)")
plt.title("První diference čtyř posloupností")
plt.grid(True)
plt.legend()
plt.show()
```

**Která posloupnost má konstantní kladnou diferenci:** modra

**U které posloupnosti se znaménko diference změní pouze jednou:** oranzova

**Která posloupnost je stále klesající:** zelena

**Která posloupnost střídavě roste a klesá:** cervena

## 2. Úloha: Monotonie z první diference

Nyní použijeme znaménko první diference přímo k určení monotonie posloupnosti

$$
a(t)=t^2-6t+4,\qquad t\geq 0.
$$

Nejprve si vypíšeme hodnoty posloupnosti a její první diference. Potom obě veličiny zobrazíme graficky.

Připomeňte si:

- $\Delta a(t)<0$ znamená pokles mezi $t$ a $t+1$,
- $\Delta a(t)>0$ znamená růst mezi $t$ a $t+1$.


```r
t8 = np.arange(0, 9)
a8 = t8**2 - 6*t8 + 4
D8 = np.diff(a8)

print(f"t     = {t8}")
print(f"a(t)  = {a8}")
print(f"Δa(t) = {D8}")
```

```r
plt.figure(figsize=(9, 5))
plt.plot(t8, a8, marker="o")
plt.axhline(0, linestyle="--")
plt.axvline(3, linestyle=":")
plt.xlabel("t")
plt.ylabel("a(t)")
plt.title("Posloupnost a(t) = t² - 6t + 4")
plt.grid(True)
plt.show()
```

```r
t8_diff = t8[:-1]

plt.figure(figsize=(9, 5))
plt.plot(t8_diff, D8, marker="o")
plt.axhline(0, linestyle="--")
plt.xlabel("t")
plt.ylabel("Δa(t)")
plt.title("První diference")
plt.grid(True)
plt.show()
```

### Co pozorujete?

a) **Pro která celá $t\geq 0$ je $\Delta a(t)<0$:** 0-2

b) **Od kterého $t$ je $\Delta a(t)>0$:** 3 a dal

c) **V kterém indexu nabývá posloupnost minima:** ve 3 hodnotu -5

d) **Na které části posloupnost klesá a na které roste:** od 0 do 3 posloupnost klesa a od 3 dal roste

e) **Jak se změna monotonie projeví v grafu první diference:** prechod pres nulu


## 3. Opakovaný posun

Dalším tématem je opakovaný posun. Pro konkrétní posloupnost porovnáme

$$
E^k a(t)
$$

a

$$
a(t+k).
$$

Zvolíme

$$
a(t)=t^2-4t+1
$$

a posun $k=3$.

Cílem není pouze ověřit rovnost, ale také si všimnout, **jak se posun projeví v tabulce hodnot a v grafu**.


```r
t = np.arange(0, 10)
k = 3

a = t**2 - 4*t + 1

# Hodnoty po třech posunech
Ek_a = a[k:]

# Přímé dosazení t+k do předpisu
t_short = t[:-k]
a_t_plus_k = (t_short + k)**2 - 4*(t_short + k) + 1

print("t =", t_short)
print("E^k a(t) =", Ek_a)
print("a(t+k)   =", a_t_plus_k)
#print("Výsledky se shodují:", np.all(Ek_a == a_t_plus_k))
```

```r
plt.figure(figsize=(9, 5))
plt.plot(t_short, Ek_a, marker="o", label="E³a(t)")
plt.plot(t_short, a_t_plus_k, marker="x", linestyle="--", label="a(t+3)")
plt.xlabel("t")
plt.ylabel("hodnota")
plt.title("Opakovaný posun")
plt.grid(True)
plt.legend()
plt.show()
```

### Co pozorujete?

a) **Shodují se hodnoty $E^k a(t)$ a $a(t+k)$:** 
ano, shoduji

b) **Co se stane, když zvýšíme hodnotu $k$:** 
nestane se nic s tvarem, jen se graf posune

c) **Jak se posun projeví v grafu posloupnosti:**
jen se posune graf

d) **Proč má posunutá posloupnost v našem výpočtu méně zobrazených hodnot:**
posune se graf, ale jinak je to furt definovane stejne daleko, takze uz nejsou data

## 4. Vyšší diference polynomiální posloupnosti

Teď budeme sledovat, co se děje při opakovaném diferencování polynomiální posloupnosti

$$
a(t)=2t^4-3t^2+1.
$$

Pomocí `np.diff` vytvoříme postupně první, druhou, třetí, čtvrtou a pátou diferenci a budeme hledat okamžik, kdy se diference stane konstantní a následně nulovou.


```r
t = np.arange(0, 11)
a = 2*t**4 - 3*t**2 + 1

Da = np.diff(a)
D2a = np.diff(a, 2)
D3a = np.diff(a, 3)
D4a = np.diff(a, 4)
D5a = np.diff(a, 5)

print(f"a(t)    = {a}")
print(f"Δa(t)   = {Da}")
print(f"Δ²a(t)  = {D2a}")
print(f"Δ³a(t)  = {D3a}")
print(f"Δ⁴a(t)  = {D4a}")
print(f"Δ⁵a(t)  = {D5a}")
```

Každá další diference obsahuje o jednu hodnotu méně. Proto má také kratší osu času.

```r
plt.figure(figsize=(9, 5))
plt.plot(t, a, marker="o")
plt.xlabel("t")
plt.ylabel("a(t)")
plt.title("Polynomiální posloupnost")
plt.grid(True)
plt.show()
```

```r
plt.figure(figsize=(9, 5))
plt.plot(t[:-1], Da, marker="o", label="Δa(t)")
plt.plot(t[:-2], D2a, marker="o", label="Δ²a(t)")
plt.plot(t[:-3], D3a, marker="o", label="Δ³a(t)")
plt.xlabel("t")
plt.ylabel("hodnota diference")
plt.title("První až třetí diference")
plt.grid(True)
plt.legend()
plt.show()
```

```r
plt.figure(figsize=(9, 5))
plt.plot(t[:-4], D4a, marker="o", label="Δ⁴a(t)")
plt.plot(t[:-5], D5a, marker="o", label="Δ⁵a(t)")
#plt.axhline(0, linestyle="--")
plt.xlabel("t")
plt.ylabel("hodnota diference")
plt.title("Čtvrtá a pátá diference")
plt.grid(True)
plt.legend()
plt.show()
```

### Co pozorujete?

a) **Která diference je poprvé konstantní:** 4.

b) **Jakou hodnotu má tato konstantní diference:** 48

c) **Která diference je poprvé identicky nulová:** 5.

d) **Jak souvisí stupeň původního polynomu s řádem konstantní a nulové diference:** stupen polynomu je roven konstantni diferenci a o jeden vetsi nez stupen je nulova diference

e) **Proč má každá další diference o jednu hodnotu méně:** nejsme schopni ziskat informace o poslednim clenu

## 5. Posloupnost z počátečních diferencí

Nyní se podíváme na opačný problém. Neznáme přímo předpis posloupnosti, ale známe její počáteční diference:

$$
a(0)=2,\qquad
\Delta a(0)=3,\qquad
\Delta^2a(0)=-1,
$$

přičemž všechny diference od třetího řádu jsou nulové.

Z těchto údajů sestavíme hodnoty posloupnosti a podíváme se, jak se informace o nulové vyšší diferenci projeví v jejím tvaru.


```r
t = np.arange(0, 11)

a0 = 2
Da0 = 3
D2a0 = -1

a = np.array([
    a0 + comb(int(tt), 1)*Da0 + comb(int(tt), 2)*D2a0
    for tt in t
])

Da = np.diff(a)
D2a = np.diff(a, 2)
D3a = np.diff(a, 3)

print("t       =", t)
print("a(t)    =", a)
print("Δa(t)   =", Da)
print("Δ²a(t)  =", D2a)
print("Δ³a(t)  =", D3a)
print("\na(10) =", a[10])
```

```r
plt.figure(figsize=(9, 5))
plt.plot(t, a, marker="o")
plt.axhline(0, linestyle="--")
plt.xlabel("t")
plt.ylabel("a(t)")
plt.title("Posloupnost rekonstruovaná z počátečních diferencí")
plt.grid(True)
plt.show()
```

])

b) **Hodnota $a(10)$:** -13

c) **Řád první nulové diference:** 3.

d) **Stupeň polynomu:** 2
